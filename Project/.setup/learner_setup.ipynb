```python
import sys
from pathlib import Path


def get_serverless_version() -> int:
    python_version = sys.version.split()[0]
    match python_version:
        case "3.10.12":
            return 1
        case "3.11.10":
            return 2
        # default to latest version
        case _:
            return 2


def get_repo_root() -> Path:
    current = Path.cwd().resolve()
    for candidate in (current, *current.parents):
        if (candidate / "notebook_utils.py").exists() or (candidate / ".git").exists():
            return candidate
    return current


repo_root = get_repo_root()
serverless_version = get_serverless_version()
print(serverless_version)
```

```python
requirements = repo_root / "Project" / "project" / str(serverless_version) / "shim.txt"
constraints = repo_root / "Project" / "project" / str(serverless_version) / "constraints.txt"

%pip install -r $requirements -c $constraints
print("The kernel has been restarted and you can now run the notebook.")
```
